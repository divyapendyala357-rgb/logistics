```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)
n = 240

regions = np.random.choice(["North", "South", "East", "West"], n)
modes = np.random.choice(["Road", "Rail", "Air", "Ship"], n, p=[0.52, 0.20, 0.18, 0.10])
volume = np.random.randint(20, 501, n)
distance = np.random.randint(50, 1801, n)

mode_factor = pd.Series(modes).map({
    "Road": 1.15, "Rail": 1.05, "Air": 0.35, "Ship": 1.65
}).to_numpy()
delivery_days = np.maximum(
    0.5,
    (distance / 450) * mode_factor
    + np.random.gamma(1.4, 0.6, n)
    + np.random.normal(0, 0.35, n)
)
cost_per_km = pd.Series(modes).map({
    "Road": 2.2, "Rail": 1.2, "Air": 7.5, "Ship": 0.8
}).to_numpy()
cost = np.maximum(
    100,
    distance * cost_per_km
    + volume * np.random.uniform(1.5, 3.5, n)
    + np.random.normal(0, 120, n)
)
threshold = {"Road": 3.8, "Rail": 4.2, "Air": 2.2, "Ship": 6.0}
status = np.where(
    delivery_days > np.array([threshold[m] for m in modes]),
    "Late", "On Time"
)

data = pd.DataFrame({
    "Shipment_ID": [f"SH{1001+i}" for i in range(n)],
    "Region": regions,
    "Transport_Mode": modes,
    "Shipment_Volume_Units": volume,
    "Distance_km": distance,
    "Delivery_Time_Days": delivery_days.round(2),
    "Transport_Cost_INR": cost.round(2),
    "Delivery_Status": status
})
data.to_csv("week3_hypothetical_logistics_dataset.csv", index=False)

print("Dataset shape:", data.shape)
print("\\nFirst five rows:\\n", data.head())
print("\\nMissing values:\\n", data.isnull().sum())
print("\\nDuplicate rows:", data.duplicated().sum())
print("\\nDescriptive statistics:\\n", data[
    ["Shipment_Volume_Units", "Distance_km",
     "Delivery_Time_Days", "Transport_Cost_INR"]
].describe().round(2))
print("\\nCorrelation matrix:\\n", data[
    ["Shipment_Volume_Units", "Distance_km",
     "Delivery_Time_Days", "Transport_Cost_INR"]
].corr().round(2))
print("\\nMode analysis:\\n", data.groupby("Transport_Mode").agg(
    Shipments=("Shipment_ID", "count"),
    Average_Delivery_Days=("Delivery_Time_Days", "mean"),
    Average_Cost_INR=("Transport_Cost_INR", "mean"),
    Late_Rate=("Delivery_Status", lambda s: (s == "Late").mean() * 100)
).round(2))

# 1. Delivery-time distribution
plt.figure(figsize=(7, 4))
plt.hist(data["Delivery_Time_Days"], bins=18, edgecolor="black")
plt.title("Distribution of Delivery Time")
plt.xlabel("Delivery time (days)")
plt.ylabel("Number of shipments")
plt.tight_layout()
plt.savefig("delivery_time_histogram.png", dpi=160)
plt.show()

# 2. Average delivery time by transport mode
data.groupby("Transport_Mode")["Delivery_Time_Days"].mean().sort_values().plot(
    kind="bar", edgecolor="black", figsize=(7, 4)
)
plt.title("Average Delivery Time by Transport Mode")
plt.xlabel("Transport mode")
plt.ylabel("Average delivery time (days)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("average_delivery_by_mode.png", dpi=160)
plt.show()

# 3. Distance versus transportation cost
plt.figure(figsize=(7, 4))
plt.scatter(data["Distance_km"], data["Transport_Cost_INR"], alpha=0.65)
plt.title("Distance and Transportation Cost")
plt.xlabel("Distance (km)")
plt.ylabel("Transportation cost (INR)")
plt.tight_layout()
plt.savefig("distance_cost_scatter.png", dpi=160)
plt.show()

# 4. Correlation heatmap
corr = data[["Shipment_Volume_Units", "Distance_km",
             "Delivery_Time_Days", "Transport_Cost_INR"]].corr()
plt.figure(figsize=(6, 4))
plt.imshow(corr, aspect="auto")
plt.colorbar(label="Correlation coefficient")
plt.xticks(range(4), ["Volume", "Distance", "Delivery time", "Cost"], rotation=25, ha="right")
plt.yticks(range(4), ["Volume", "Distance", "Delivery time", "Cost"])
plt.title("Correlation Matrix of Numeric Variables")
plt.tight_layout()
plt.savefig("correlation_heatmap.png", dpi=160)
plt.show()

# 5. Late delivery rate by region
data.groupby("Region")["Delivery_Status"].apply(
    lambda s: (s == "Late").mean() * 100
).plot(kind="bar", edgecolor="black", figsize=(7, 4))
plt.title("Late Delivery Rate by Region")
plt.xlabel("Region")
plt.ylabel("Late shipments (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("late_rate_by_region.png", dpi=160)
plt.show()

# 6. Cost spread by transport mode
data.boxplot(column="Transport_Cost_INR", by="Transport_Mode", grid=False, figsize=(7, 4))
plt.suptitle("")
plt.title("Transportation Cost Spread by Mode")
plt.xlabel("Transport mode")
plt.ylabel("Transportation cost (INR)")
plt.tight_layout()
plt.savefig("cost_boxplot_by_mode.png", dpi=160)
plt.show()
```
